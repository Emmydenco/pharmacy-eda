### Plot for Average age per drug

```python

sns.set_theme(style="whitegrid")


plt.figure(figsize=(10, 6))
ax = sns.barplot(data=df, x='Drug', y='Age', errorbar=None)

plt.xlabel('Drug', fontsize=12)
plt.ylabel('Average Age', fontsize=12)
plt.title('Average Age of Patients per Drug', fontsize=14, fontweight='bold')

# Rotate the x-axis labels if drug names are long or overlapping
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("images/avg_age_per_drug.png", dpi=150, bbox_inches="tight")

plt.show()
```

"Mean age is lowest for drugA (35.7) and highest for drugB (61.0), while the other drugs cluster between 40 and 45

```python

sns.set_theme(style="whitegrid")

plt.figure(figsize=(9, 5))

sns.histplot(data=df, x='Na_to_K', kde=True, color='teal', bins=20)

plt.xlabel('Sodium-to-Potassium Ratio (Na_to_K)', fontsize=12)
plt.ylabel('Count / Frequency', fontsize=12)
plt.title('Distribution of Na_to_K Ratios in Patients', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.savefig("images/na_to_k_distribution.png", dpi=150, bbox_inches="tight")

plt.show()
```
